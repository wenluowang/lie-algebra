## 1. Basic Lie Algebra Concepts

### 1.1 Creating a Simple Lie Algebra

Let's start by creating the simplest non-trivial Lie algebra: sl(2), the special linear Lie algebra.

```python
# Create sl(2) using Cartan type notation ['A', 1]
L = LieAlgebra(QQ, cartan_type=['A', 1])
print(f"Lie algebra: {L}")
print(f"\nCartan type: {L.cartan_type()}")
print(f"\nBase ring: {L.base_ring()}")
```

### 1.2 Basis Elements

Every Lie algebra has a basis. For simple Lie algebras, we have:
- e_i: raising operators
- f_i: lowering operators
- h_i: Cartan subalgebra elements

```python
# Get the generators
e1 = L.e(1)
f1 = L.f(1)
h1 = L.h(1)

print("Generators of sl(2):")
print(f"e1 = {e1}")
print(f"f1 = {f1}")
print(f"h1 = {h1}")
```

### 1.3 Lie Bracket

The defining operation of a Lie algebra is the Lie bracket [x, y].

```python
# Compute Lie brackets
bracket_ef = e1.bracket(f1)
bracket_he = h1.bracket(e1)
bracket_hf = h1.bracket(f1)

print("Lie brackets in sl(2):")
print(f"[e1, f1] = {bracket_ef}")
print(f"[h1, e1] = {bracket_he}")
print(f"[h1, f1] = {bracket_hf}")
```

### 1.4 Verifying the Jacobi Identity

One of the key properties of a Lie algebra is the Jacobi identity:
[x, [y, z]] + [y, [z, x]] + [z, [x, y]] = 0

```python
# Verify Jacobi identity for e1, f1, h1
x, y, z = e1, f1, h1

term1 = x.bracket(y.bracket(z))
term2 = y.bracket(z.bracket(x))
term3 = z.bracket(x.bracket(y))

jacobi_sum = term1 + term2 + term3

print("Jacobi identity verification:")
print(f"[e1, [f1, h1]] = {term1}")
print(f"[f1, [h1, e1]] = {term2}")
print(f"[h1, [e1, f1]] = {term3}")
print(f"\nSum (should be 0): {jacobi_sum}")
print(f"Is zero? {jacobi_sum == 0}")
```

## 2. Root Systems

Root systems are fundamental to understanding the structure of Lie algebras.

```python
# Create a root system for type A2 (sl(3))
R = RootSystem(['A', 2])
print(f"Root system: {R}")

# Get the root lattice
root_lattice = R.root_lattice()
print(f"\nRoot lattice: {root_lattice}")

# Get positive roots
positive_roots = list(root_lattice.positive_roots())
print(f"\nPositive roots: {positive_roots}")

# Get simple roots
simple_roots = root_lattice.simple_roots()
print(f"\nSimple roots: {simple_roots}")
```

### 2.1 Cartan Matrix

The Cartan matrix encodes the inner products of simple roots.

```python
# Get the Cartan matrix
cartan_matrix = R.cartan_matrix()
print("Cartan matrix for A2:")
print(cartan_matrix)
```

### 2.2 Dynkin Diagram

The Dynkin diagram is a graphical representation of the Cartan matrix.

```python
# Display Dynkin diagram information
dynkin = R.dynkin_diagram()
print("Dynkin diagram for A2:")
print(dynkin)
```

## 3. Classical Lie Algebras

Let's explore the classical Lie algebras: A_n, B_n, C_n, D_n.

```python
# Type A_n: sl(n+1)
A3 = LieAlgebra(QQ, cartan_type=['A', 3])
print(f"Type A3 (sl(4)): {A3}")
print(f"Dimension: {A3.dimension()}")

# Type B_n: so(2n+1)
B2 = LieAlgebra(QQ, cartan_type=['B', 2])
print(f"\nType B2 (so(5)): {B2}")
print(f"Dimension: {B2.dimension()}")

# Type C_n: sp(2n)
C2 = LieAlgebra(QQ, cartan_type=['C', 2])
print(f"\nType C2 (sp(4)): {C2}")
print(f"Dimension: {C2.dimension()}")

# Type D_n: so(2n)
D3 = LieAlgebra(QQ, cartan_type=['D', 3])
print(f"\nType D3 (so(6)): {D3}")
print(f"Dimension: {D3.dimension()}")
```

## 4. Matrix Representation of sl(2)

Let's work with the concrete matrix representation of sl(2).

```python
# Define sl(2) using 2x2 matrices
E = matrix(QQ, [[0, 1], [0, 0]])  # Raising operator
F = matrix(QQ, [[0, 0], [1, 0]])  # Lowering operator
H = matrix(QQ, [[1, 0], [0, -1]]) # Diagonal element

print("Matrix representation of sl(2):")
print(f"\nE (raising) = \n{E}")
print(f"\nF (lowering) = \n{F}")
print(f"\nH (diagonal) = \n{H}")

# Verify these form a basis of sl(2)
print("\nTrace checks (all should be 0 for sl(2)):")
print(f"trace(E) = {E.trace()}")
print(f"trace(F) = {F.trace()}")
print(f"trace(H) = {H.trace()}")
```

```python
# Compute Lie brackets [X, Y] = XY - YX
def lie_bracket(X, Y):
    return X*Y - Y*X

print("Lie brackets:")
print(f"\n[E, F] = \n{lie_bracket(E, F)}")
print(f"\n[H, E] = \n{lie_bracket(H, E)}")
print(f"\n[H, F] = \n{lie_bracket(H, F)}")

# Verify: [E,F] = H, [H,E] = 2E, [H,F] = -2F
print("\nVerification:")
print(f"[E, F] == H: {lie_bracket(E, F) == H}")
print(f"[H, E] == 2E: {lie_bracket(H, E) == 2*E}")
print(f"[H, F] == -2F: {lie_bracket(H, F) == -2*F}")
```

## 5. Weight Spaces and Representations

Understanding representations is crucial for applications of Lie algebras.

```python
# Work with the weight lattice for type A2
R = RootSystem(['A', 2])
P = R.weight_lattice()

print(f"Weight lattice: {P}")

# Fundamental weights
fw = P.fundamental_weights()
print(f"\nFundamental weights: {fw}")

# Simple roots in the weight lattice
sr = P.simple_roots()
print(f"\nSimple roots: {sr}")
```
